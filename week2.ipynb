```python
df = pd.read_csv('placement_predict_50k.csv')
fig1 = plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
plt.title("Missingness Heatmap - Placement Prediction Dataset")
plt.xlabel("Columns")
plt.tight_layout()
plt.show()
```

```python
num_cols = [
    'CGPA',
    'AttendancePercent',
    'AptitudeTestScore',
    'CodingTestScore',
    'SoftSkillsRating',
    'PlacementPackage'   # Replace with the correct column name if different
]

fig2, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    if col in df.columns:
        sns.histplot(df[col].dropna(), bins=30, kde=True,
                     ax=axes[i], color='steelblue')
        axes[i].set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

```python
pdf = PdfPages('EDA_Report.pdf')

pdf.savefig(fig1)   # Page 1: Missingness Heatmap
pdf.savefig(fig2)   # Page 2: Six Histograms
```

```python
fig5, axes5 = plt.subplots(2, 3, figsize=(15, 8))
axes5 = axes5.flatten()

for i, col in enumerate(num_cols):
    if col in df.columns:
        sns.boxplot(y=df[col], ax=axes5[i], color='lightcoral')
        axes5[i].set_title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
fig4 = plt.figure(figsize=(10, 8))

numeric_df = df.select_dtypes(include=['int64', 'float64'])
corr_matrix = numeric_df.corr()

sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, linewidths=0.5)
plt.title('Correlation Heatmap - Numerical Features')
plt.tight_layout()
plt.show()
```
